````python
EDGEWORK_INSTRUCTIONS = """
You are the intelligence behind the Edgework Translation Desk.

PURPOSE

Edgework helps people understand technical, legal, policy, research,
and other complex information surrounding technology.

The user's input may be:
- a headline
- a news excerpt
- a paragraph
- a policy statement
- legislation
- technical documentation
- a research finding
- a social media post
- a question
- or another piece of information they encountered.

The user may not know what type of source they are providing.

YOUR JOB

Do more than summarize.

Help the user understand what they encountered and why it matters.

For every request:

1. Explain the information in clear, accessible language.
2. Preserve important nuance and qualifications.
3. Distinguish between:
   - what the provided material explicitly says
   - what can reasonably be inferred
   - what remains uncertain or unsupported
4. Explain the relevant context.
5. Identify who or what may be affected.
6. Identify important limitations, ambiguity, or missing information.
7. When useful, suggest questions the user could ask next.
8. Do not invent facts, requirements, statistics, or conclusions.
9. When discussing laws or regulations, explain rather than provide
   individualized legal advice.
10. Prefer authoritative and primary sources when available.
11. Clearly distinguish information from the user's material,
    retrieved sources, and your own explanation.

SOURCE PRINCIPLES

Prefer sources in roughly this order when appropriate:

1. Legislation and primary legal materials
2. Government agencies and regulatory bodies
3. Official standards organizations
4. Official technical documentation
5. Research institutions and peer-reviewed research
6. Reputable secondary reporting

Do not treat all sources as equally authoritative.

If sources disagree, say so.

If there is insufficient evidence to answer confidently,
say that rather than filling the gap with speculation.

STYLE

Write for an intelligent person who is not necessarily an expert.

Be:
- clear
- thoughtful
- concise
- curious
- nuanced
- non-alarmist
- non-condescending

Do not unnecessarily use jargon.

CORE PRINCIPLE

The goal is not to remove complexity.

The goal is to make complexity navigable.
"""
````

````python
def build_user_input(material="", question=""):
    """
    Create a standardized input for the Edgework Translation Desk.
    """

    if material.strip():
        return f"""
The user encountered the following material:

--- BEGIN MATERIAL ---

{material}

--- END MATERIAL ---

The user's question is:

{question}
"""

    return f"""
The user's question is:

{question}
"""
````

````python
def ask_edgework_with_web(user_input):
    """
    Ask Edgework to answer a question using live web retrieval
    when current or external information is needed.
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=f"""
{EDGEWORK_INSTRUCTIONS}

The user has provided the following material or question:

{user_input}

Use Google Search when current, changing, or externally
verifiable information is needed.

Prioritize authoritative sources such as:

- legislation
- government agencies
- regulatory bodies
- standards organizations
- official technical documentation
- research institutions
- reputable reporting

Use retrieved information to verify claims and provide
relevant context.

Do not assume that a search result is authoritative simply
because it appears near the top of the search results.

If reliable sources disagree, explain that disagreement.
""",
        config=types.GenerateContentConfig(
            temperature=0.2,
            max_output_tokens=2500,
            tools=[
                types.Tool(
                    google_search=types.GoogleSearch()
                )
            ],
        ),
    )

    return response.text
````

````python
# cell 13

def ask_edgework_with_url(user_input, url):
    """
    Ask Edgework to analyze user-provided material
    using a specific public URL as additional source context.
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",

        contents=f"""
{EDGEWORK_INSTRUCTIONS}

The user provided the following material or question:

{user_input}

The user also provided this source URL:

{url}

Use the contents of the provided URL to verify, contextualize,
or explain the user's material.

Treat the URL as source material, not automatically as fact.

Clearly distinguish:
- what the source explicitly states
- what can reasonably be inferred
- what remains uncertain

Prefer the source's own language and context over speculation.
""",

        config=types.GenerateContentConfig(
            temperature=0.2,
            max_output_tokens=2500,

            tools=[
                types.Tool(
                    url_context=types.UrlContext()
                )
            ],
        ),
    )

    return response.text
````

````python
# cell 20

def identify_relevant_sources(user_input):
    """
    Determine which Edgework trusted sources are potentially
    relevant to a user's material/question.
    """

    source_descriptions = ""

    for key, source in TRUSTED_SOURCES.items():
        source_descriptions += f"""
SOURCE: {key}
Organization: {source["organization"]}
Category: {source["category"]}
Description: {source["description"]}
Topics: {", ".join(source["topics"])}
URLs: {", ".join(source["urls"])}
---
"""

    prompt = f"""
You are helping the Edgework Translation Desk determine
which trusted sources may be relevant to a user's question.

USER INPUT:
{user_input}

AVAILABLE TRUSTED SOURCES:
{source_descriptions}

Identify the sources that are genuinely relevant to the
user's question.

Do not select a source merely because it is available.

Return ONLY a Python-style list containing the source keys.

For example:

["NIST", "EU_AI_ACT"]

If none are clearly relevant, return:

[]
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=200
        )
    )

    return response.text.strip()
````

````python
# cell 22

def parse_source_selection(selection_text):
    """
    Convert Gemini's source-selection response into
    valid trusted-source keys.
    """

    import ast

    try:
        selected = ast.literal_eval(selection_text)

        if not isinstance(selected, list):
            return []

        valid_sources = [
            source for source in selected
            if source in TRUSTED_SOURCES
        ]

        return valid_sources

    except Exception:
        return []
````

````python
# cell 23
def get_source_urls(source_keys):
    """
    Return the URLs associated with selected trusted sources.
    """

    urls = []

    for key in source_keys:
        source = TRUSTED_SOURCES.get(key)

        if source:
            urls.extend(source["urls"])

    return urls
````

````python
# cell 24 create the retrieval decision

def determine_retrieval_need(user_input):
    """
    Determine what type of information Edgework needs
    before answering the user's question.

    Possible retrieval modes:
        - NONE
        - TRUSTED_SOURCE
        - USER_URL
    """

    prompt = f"""
You are the retrieval decision layer for the Edgework Translation Desk.

Your job is NOT to answer the user's question.

Your job is to determine whether additional source material
is needed before answering.

USER INPUT:
{user_input}

Choose exactly ONE retrieval mode:

NONE
Use this when the question can reasonably be answered
without consulting an external source.

TRUSTED_SOURCE
Use this when authoritative government, regulatory,
standards, legislative, technical, or research material
would materially improve the answer.

USER_URL
Use this when the user has explicitly provided a URL
that should be analyzed.

Important:
- Do not choose TRUSTED_SOURCE merely because the topic
  involves technology.
- Questions involving laws, regulations, government policy,
  standards, compliance, or current rules should generally
  use TRUSTED_SOURCE.
- Questions about current events should generally require
  current external information, but live web search is
  currently unavailable in this prototype.
- Do not invent URLs.

Return ONLY one of:

NONE
TRUSTED_SOURCE
USER_URL
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=50
        )
    )

    return response.text.strip()
````

````python
# cell 28

def determine_edgework_route(request):
    """
    Determine how Edgework should handle a user request.
    """

    if has_user_url(request):
        return "USER_URL"

    user_input = build_user_input(
        material=request["material"],
        question=request["question"]
    )

    return determine_retrieval_need(user_input)
````

````python
# cell 30
import json


def identify_relevant_sources(user_input):
    """
    Identify which trusted Edgework sources are relevant
    to the user's question.

    Returns structured JSON.
    """

    source_descriptions = ""

    for key, source in TRUSTED_SOURCES.items():

        source_descriptions += f"""
SOURCE KEY: {key}
Organization: {source["organization"]}
Source Type: {source["source_type"]}
Authority Level: {source["authority_level"]}
Description: {source["description"]}
Topics: {", ".join(source["topics"])}
URLs: {", ".join(source["urls"])}
---
"""

    prompt = f"""
You are the source-selection layer for the Edgework Translation Desk.

Your job is NOT to answer the user's question.

Your job is to identify which trusted sources would materially
improve the accuracy or context of the answer.

USER INPUT:
{user_input}

AVAILABLE TRUSTED SOURCES:

{source_descriptions}

Rules:

1. Select only sources that are genuinely relevant.
2. Do not select a source simply because it exists.
3. If the question concerns law, regulation, compliance,
   government policy, or standards, prioritize primary
   institutional sources.
4. If the question is conceptual and does not require
   current information, fewer sources may be appropriate.
5. Never invent a source.
6. It is acceptable to select zero sources.

Return ONLY valid JSON in this exact format:

{{
    "sources": [
        {{
            "key": "NIST",
            "reason": "Brief explanation of why this source is relevant."
        }}
    ]
}}

If no source is relevant:

{{
    "sources": []
}}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=500
        )
    )

    try:
        return json.loads(response.text)

    except Exception:
        print("Could not parse source-selection response.")
        print(response.text)

        return {
            "sources": []
        }
````

````python
# cell 34 validate the sources

def validate_source_selection(selection):
    """
    Validate Gemini's selected sources against the
    Edgework trusted-source registry.
    """

    valid_sources = []

    for item in selection.get("sources", []):

        key = item.get("key")

        if key in TRUSTED_SOURCES:
            valid_sources.append({
                "key": key,
                "reason": item.get("reason", "")
            })

    return valid_sources
````

````python
# cell 33 retrieve the urls

def get_selected_source_urls(valid_sources):
    """
    Convert validated source selections into
    source URLs and metadata.
    """

    selected = []

    for item in valid_sources:

        key = item["key"]
        source = TRUSTED_SOURCES[key]

        for url in source["urls"]:

            selected.append({
                "key": key,
                "organization": source["organization"],
                "source_type": source["source_type"],
                "authority_level": source["authority_level"],
                "url": url,
                "reason": item["reason"]
            })

    return selected
````

````python
# cell 36: build the actual master function

def edgework_translate(
    material="",
    question="",
    url=None
):
    """
    Main Edgework Translation Desk pipeline.

    Inputs:
        material - text the user encountered
        question - what the user wants to understand
        url      - optional URL provided by the user

    Returns:
        Edgework-generated explanation.
    """

    # ------------------------------------------------------------
    # STEP 1: Create structured request
    # ------------------------------------------------------------

    request = create_edgework_request(
        material=material,
        question=question,
        url=url
    )


    # ------------------------------------------------------------
    # STEP 2: Build user input for the model
    # ------------------------------------------------------------

    user_input = build_user_input(
        material=request["material"],
        question=request["question"]
    )


    # ------------------------------------------------------------
    # STEP 3: Determine routing
    # ------------------------------------------------------------

    route = determine_edgework_route(request)

    print(f"Edgework route: {route}")


    # ------------------------------------------------------------
    # STEP 4: User-provided URL
    # ------------------------------------------------------------

    if route == "USER_URL":

        return ask_edgework_with_url(
            user_input=user_input,
            url=request["url"]
        )


    # ------------------------------------------------------------
    # STEP 5: No external source required
    # ------------------------------------------------------------

    if route == "NONE":

        return ask_edgework_with_sources(
            user_input=user_input,
            selected_sources=[]
        )


    # ------------------------------------------------------------
    # STEP 6: Trusted-source retrieval
    # ------------------------------------------------------------

    if route == "TRUSTED_SOURCE":

        selection = identify_relevant_sources(user_input)

        valid_sources = validate_source_selection(
            selection
        )

        selected_sources = get_selected_source_urls(
            valid_sources
        )

        print("Selected trusted sources:")

        for source in selected_sources:
            print(
                f"- {source['organization']} "
                f"({source['source_type']})"
            )

        return ask_edgework_with_sources(
            user_input=user_input,
            selected_sources=selected_sources
        )


    # ------------------------------------------------------------
    # STEP 7: Safety fallback
    # ------------------------------------------------------------

    return """
Edgework could not determine the appropriate retrieval
path for this question. Please provide additional context.
"""
````

````python
# cell 50 new information need classifier

def classify_information_need(user_input):
    """
    Determine what kind of information Edgework needs
    to answer the user's question responsibly.
    """

    prompt = f"""
You are the information-routing layer for the
Edgework Translation Desk.

Your job is NOT to answer the user's question.

Your job is to determine what kind of information
Edgework needs in order to answer it accurately.

USER INPUT:

{user_input}

Choose exactly ONE category:

CONCEPTUAL
Use when the user primarily needs an explanation
of a concept, term, or idea and authoritative retrieval
is not necessary.

REGULATORY
Use when the answer depends on formal regulations,
regulatory frameworks, compliance requirements,
or government rules.

LEGAL
Use when the user is asking about legality, legal
rights, liability, legal obligations, or interpretation
of law.

POLICY
Use when the question concerns government policy,
institutional governance, or public policy.

EMPIRICAL
Use when the question asks what research, studies,
or evidence show.

TECHNICAL
Use when the question requires technical documentation
or explanation of how a specific technology or system works.

CURRENT_EVENT
Use when the question depends on a recent development,
announcement, event, or change.

MIXED
Use only when two or more categories are genuinely
necessary to answer the question.

IMPORTANT:

The user's wording may be conceptual even when the
underlying information is regulatory, legal, empirical,
or technical.

Classify based on the information required to answer
the question, NOT simply on the wording "what does this mean?"

Examples:

"What is an AI agent?"
→ CONCEPTUAL

"What does the EU AI Act require?"
→ REGULATORY

"Can my employer legally use AI to screen applicants?"
→ LEGAL

"Does AI actually increase productivity?"
→ EMPIRICAL

"How does this API authenticate requests?"
→ TECHNICAL

"Did the government change its AI rules this week?"
→ CURRENT_EVENT

"What does AI governance mean, and is it legally required?"
→ MIXED

Return ONLY valid JSON:

{{
    "information_need": "CONCEPTUAL",
    "reason": "Brief explanation."
}}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=250
        )
    )

    # ------------------------------------------------------------
    # Robust JSON extraction
    # ------------------------------------------------------------

    text = response.text.strip()

    try:

        return json.loads(text)

    except json.JSONDecodeError:

        # Try extracting the JSON object if Gemini
        # wrapped it in markdown or extra text.

        start = text.find("{")
        end = text.rfind("}")

        if start != -1 and end != -1:

            try:

                return json.loads(
                    text[start:end + 1]
                )

            except json.JSONDecodeError:
                pass

        print("Could not parse information-need classification.")
        print(text)

        return {
            "information_need": "CONCEPTUAL",
            "reason": "Fallback classification."
        }
````

````python
# cell 54 helper fix json parsing everywhere

def parse_json_response(text, fallback=None):
    """
    Robustly parse JSON returned by Gemini.

    Handles:
    - pure JSON
    - JSON surrounded by whitespace
    - JSON inside markdown code fences
    - JSON surrounded by explanatory text
    """

    if fallback is None:
        fallback = {}

    text = text.strip()

    # First attempt: direct parsing
    try:
        return json.loads(text)

    except json.JSONDecodeError:
        pass


    # Second attempt: remove markdown code fences
    cleaned = (
        text
        .replace("```json", "")
        .replace("```", "")
        .strip()
    )

    try:
        return json.loads(cleaned)

    except json.JSONDecodeError:
        pass


    # Third attempt: extract the outermost JSON object
    start = cleaned.find("{")
    end = cleaned.rfind("}")

    if start != -1 and end != -1:

        try:

            return json.loads(
                cleaned[start:end + 1]
            )

        except json.JSONDecodeError:
            pass


    print("Unable to parse JSON response:")
    print(text)

    return fallback
````

````python
# cell 59 build the new source selector

def select_edgework_sources(
    user_input,
    information_need
):
    """
    Select the minimum number of appropriate trusted sources
    from the source pool allowed for the information need.
    """

    candidate_sources = get_sources_for_information_need(
        information_need
    )

    # No retrieval needed
    if not candidate_sources:
        return {
            "sources": []
        }

    # Build a concise description of the candidate pool
    source_descriptions = ""

    for key, source in candidate_sources.items():

        source_descriptions += f"""
SOURCE KEY: {key}
Organization: {source["organization"]}
Category: {source["category"]}
Source Type: {source["source_type"]}
Authority: {source["authority_level"]}
Description: {source["description"]}
Topics: {", ".join(source["topics"])}
"""

    prompt = f"""
You are the source-selection layer of the
Edgework Translation Desk.

The user has already been classified as requiring:

{information_need}

Your task is to select the MINIMUM number of trusted
sources necessary to answer the user's question well.

USER INPUT:
{user_input}

CANDIDATE SOURCES:
{source_descriptions}

Rules:

1. Select only sources that are directly relevant.
2. Do not select a source merely because it is generally
   related to artificial intelligence.
3. Prefer the most authoritative source for the specific issue.
4. Select the minimum number of sources necessary.
5. If one source is sufficient, select one source.
6. Do not invent sources.
7. Only use SOURCE KEY values that appear in the candidate list.
8. If no source is genuinely useful, return an empty list.

Return ONLY valid JSON in this format:

{{
    "sources": [
        {{
            "key": "SOURCE_KEY",
            "reason": "Brief explanation of why this source is needed."
        }}
    ]
}}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=500
        )
    )

    result = parse_json_response(
        response.text,
        fallback={"sources": []}
    )

    return result
````

````python
# cell 60 add a safety check

def validate_edgework_sources(
    source_selection,
    information_need
):
    """
    Validate Gemini's source selections against the
    source pool that Python authorized.
    """

    allowed_sources = get_sources_for_information_need(
        information_need
    )

    validated = []

    for item in source_selection.get("sources", []):

        key = item.get("key")

        if key in allowed_sources:

            validated.append({
                "key": key,
                "reason": item.get(
                    "reason",
                    "Relevant trusted source."
                ),
                "source": allowed_sources[key]
            })

    return validated
````

````python
# cell 65 automatically retrieve the NIST page

def retrieve_webpage(url):
    """
    Retrieve text from a public webpage.
    """

    response = requests.get(
        url,
        timeout=20,
        headers={
            "User-Agent": "Mozilla/5.0"
        }
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    # Remove elements that aren't useful source content
    for element in soup([
        "script",
        "style",
        "nav",
        "footer",
        "header"
    ]):
        element.decompose()

    text = soup.get_text(
        separator=" ",
        strip=True
    )

    return text
````

````python
# cell 67 clean the retrieved source

import re

def clean_source_text(text):
    """
    Clean retrieved webpage text before it enters
    the Edgework retrieval pipeline.
    """

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove common webpage boilerplate
    boilerplate = [
        "Skip to main content",
        "Was this page helpful?",
    ]

    for phrase in boilerplate:
        text = text.replace(phrase, "")

    # Remove repeated whitespace again
    text = re.sub(r"\s+", " ", text)

    return text.strip()
````

````python
# cell 68 turn the source into chunks

def chunk_text(
    text,
    chunk_size=1200,
    overlap=200
):
    """
    Split source text into overlapping chunks.
    """

    chunks = []

    start = 0
    text_length = len(text)

    while start < text_length:

        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
````

````python
# cell 69 improve the source cleaning
def clean_nist_source(text):
    """
    Clean NIST webpage text by removing common
    navigation and webpage boilerplate.
    """

    # Remove common NIST webpage boilerplate
    phrases_to_remove = [
        "Skip to main content",
        "Official websites use .gov",
        "A .gov website belongs to an official government organization in the United States.",
        "Secure .gov websites use HTTPS",
        "A lock ( Lock A locked padlock ) or https:// means you’ve safely connected to a .gov website.",
        "Share sensitive information only on official, secure websites.",
        "Quick Links",
        "Contact Us",
        "Was this page helpful?"
    ]

    cleaned = text

    for phrase in phrases_to_remove:
        cleaned = cleaned.replace(phrase, "")

    # Normalize whitespace
    cleaned = re.sub(r"\s+", " ", cleaned)

    return cleaned.strip()
````

````python
# cell 70 Better Chunking

import re

def chunk_text_by_sentence(
    text,
    max_characters=1200
):
    """
    Create retrieval chunks using sentence boundaries
    rather than cutting through words or sentences.
    """

    sentences = re.split(
        r'(?<=[.!?])\s+',
        text
    )

    chunks = []
    current_chunk = ""

    for sentence in sentences:

        sentence = sentence.strip()

        if not sentence:
            continue

        # If adding the sentence stays within our target size
        if len(current_chunk) + len(sentence) + 1 <= max_characters:

            if current_chunk:
                current_chunk += " "

            current_chunk += sentence

        else:

            if current_chunk:
                chunks.append(current_chunk)

            current_chunk = sentence

    # Add final chunk
    if current_chunk:
        chunks.append(current_chunk)

    return chunks
````

````python
# cell 73 - Create our retrieval index

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(
    stop_words="english"
)

chunk_vectors = vectorizer.fit_transform(
    nist_chunks
)

print(
    f"Indexed {len(nist_chunks)} NIST chunks."
)

print(
    f"Vector dimensions: {chunk_vectors.shape}"
)
````

````python
# cell 74 build the retrieval function

def retrieve_relevant_chunks(
    question,
    chunks,
    chunk_vectors,
    vectorizer,
    top_k=2
):
    """
    Retrieve the most relevant source chunks
    for a user's question.
    """

    question_vector = vectorizer.transform(
        [question]
    )

    similarities = cosine_similarity(
        question_vector,
        chunk_vectors
    )[0]

    ranked_indices = similarities.argsort()[::-1]

    results = []

    for index in ranked_indices[:top_k]:

        results.append({
            "chunk_id": int(index),
            "score": float(similarities[index]),
            "text": chunks[index]
        })

    return results
````

````python
# cell 79 D - generate
import time

def generate_edgework_translation(
    user_question,
    retrieved_context,
    max_retries=3
):
    """
    Generate an Edgework translation using Gemini
    with automatic retries for temporary service errors.
    """

    prompt = f"""
You are the Edgework Translation Desk.

Your purpose is to make technical, legal, policy,
and emerging-technology information understandable
to non-experts.

The user encountered something they do not understand.

Answer their question in clear, accessible language.

IMPORTANT:
- Use the retrieved source material below as evidence.
- Do not invent facts that are not supported by the source.
- Distinguish what the source explicitly says from reasonable
  interpretation.
- If the source does not contain enough information, say so.
- Preserve important nuance.
- Do not provide legal advice.
- Do not pretend that a voluntary framework is a law.
- Explain why the information matters when appropriate.

USER QUESTION:
{user_question}

RETRIEVED SOURCE MATERIAL:
{retrieved_context}

Provide a concise but useful Edgework Translation.
"""

    for attempt in range(max_retries):

        try:

            response = client.models.generate_content(
                model="gemini-3.7-flash",
                contents=prompt
            )

            return response.text

        except Exception as e:

            error_text = str(e)

            # Retry temporary Gemini availability errors
            if "503" in error_text or "UNAVAILABLE" in error_text:

                if attempt < max_retries - 1:

                    wait_time = 5 * (attempt + 1)

                    print(
                        f"Gemini is temporarily unavailable. "
                        f"Retrying in {wait_time} seconds..."
                    )

                    time.sleep(wait_time)

                else:

                    raise RuntimeError(
                        "Gemini remained unavailable after "
                        f"{max_retries} attempts."
                    ) from e

            else:
                # Don't hide other errors
                raise
````
